```python
import time
import pandas as pd
from datetime import datetime
from zoneinfo import ZoneInfo

KST = ZoneInfo('Asia/Seoul')
CLOSE_CUTOFF_HOUR = 16   # 장중이면 직전 영업일 사용

# KRX 섹터지수 17개 (빼려면 줄 삭제)
SECTORS = {
    '5044':'반도체', '5043':'자동차', '5048':'에너지화학', '5045':'헬스케어',
    '5064':'정보기술', '5046':'은행', '5054':'증권', '5056':'보험',
    '5049':'철강', '5052':'건설', '5057':'운송', '5051':'방송통신',
    '5063':'K콘텐츠', '5061':'경기소비재', '5062':'필수소비재',
    '5055':'기계장비', '5065':'유틸리티',
}

def sector_returns():
    now = datetime.now(KST)
    start = (now - pd.Timedelta(days=20)).strftime('%Y%m%d')
    today = now.strftime('%Y%m%d')
    rows = []
    for code, name in SECTORS.items():
        try:
            df = stock.get_index_ohlcv(start, today, code)
            if now.hour < CLOSE_CUTOFF_HOUR:          # 장중이면 오늘(미확정) 제외
                df = df[df.index.date < now.date()]
            ret = df['종가'].pct_change().iloc[-1] * 100
            rows.append({'섹터': name,
                         '종가': round(df['종가'].iloc[-1], 2),
                         '등락률(%)': round(ret, 2),
                         '기준일': df.index[-1].strftime('%Y-%m-%d')})
        except Exception as e:
            print(f'  ⚠ {name}({code}) 실패: {str(e)[:50]}')
        time.sleep(0.4)                                # KRX 과부하 방지
    return pd.DataFrame(rows).sort_values('등락률(%)', ascending=False)

sec_df = sector_returns()
print(sec_df.to_markdown(index=False))
```

```python
KST = ZoneInfo('Asia/Seoul'); CLOSE_CUTOFF_HOUR = 16   # STEP4 안 돌렸어도 되게 재선언

def base_date():
    """완료영업일(YYYYMMDD). 장중이면 직전 영업일."""
    now = datetime.now(KST)
    cal = stock.get_index_ohlcv((now - pd.Timedelta(days=20)).strftime('%Y%m%d'),
                                 now.strftime('%Y%m%d'), '1001')   # 코스피 달력
    if now.hour < CLOSE_CUTOFF_HOUR:
        cal = cal[cal.index.date < now.date()]
    return cal.index[-1].strftime('%Y%m%d')

d = base_date()
print('기준일:', datetime.strptime(d, '%Y%m%d').strftime('%Y-%m-%d'), '\n')

want = ['개인', '외국인', '기관합계']
rows = []
for mkt, label in [('KOSPI', '코스피'), ('KOSDAQ', '코스닥')]:
    inv = stock.get_market_trading_value_by_investor(d, d, mkt)   # 그 날 투자자별
    row = {'시장': label}
    for x in want:
        row[f'{x}(억)'] = f"{inv.loc[x, '순매수']/1e8:+,.0f}" if x in inv.index else '-'
    rows.append(row)

print('■ 수급 (순매수, 억원)')
print(pd.DataFrame(rows).to_markdown(index=False))

# 라벨 확인용: 코스피 전체 투자자별 순매수 (라벨 맞는지 + 연기금/금융투자 세부 확인)
print('\n[참고] 코스피 투자자별 순매수(억):')
print((stock.get_market_trading_value_by_investor(d, d, 'KOSPI')['순매수'] / 1e8).round(0))
```

```python
import time
import pandas as pd
from datetime import datetime
from zoneinfo import ZoneInfo
KST = ZoneInfo('Asia/Seoul'); CLOSE_CUTOFF_HOUR = 16

BANKS = {
    '105560':'KB금융', '055550':'신한지주', '086790':'하나금융지주',
    '316140':'우리금융지주', '024110':'기업은행', '138930':'BNK금융지주',
    '139130':'iM금융지주', '175330':'JB금융지주', '323410':'카카오뱅크',
}

def stock_returns(tickers):
    now = datetime.now(KST)
    start = (now - pd.Timedelta(days=20)).strftime('%Y%m%d')
    today = now.strftime('%Y%m%d')
    rows = []
    for code, name in tickers.items():
        try:
            df = stock.get_market_ohlcv(start, today, code)
            if now.hour < CLOSE_CUTOFF_HOUR:               # 장중이면 오늘(미확정) 제외
                df = df[df.index.date < now.date()]
            last = df.iloc[-1]
            rows.append({'종목': name,
                         '종가': f"{last['종가']:,.0f}",
                         '등락률(%)': round(last['등락률'], 2),   # 이미 % 단위 (×100 안 함)
                         '기준일': df.index[-1].strftime('%Y-%m-%d')})
        except Exception as e:
            print(f'  ⚠ {name}({code}) 실패: {str(e)[:50]}')
        time.sleep(0.4)
    return pd.DataFrame(rows)        # 입력 순서 유지 (고정 peer group)

bank_df = stock_returns(BANKS)
print(bank_df.to_markdown(index=False))
```
