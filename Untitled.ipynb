```python
import time
import pandas as pd
from datetime import datetime
from zoneinfo import ZoneInfo
from IPython.display import HTML, display
import html

KST = ZoneInfo('Asia/Seoul')
CLOSE_CUTOFF_HOUR = 16   # 이 시각 전 실행이면 직전 영업일 사용

INDICES = {'1001': '코스피', '2001': '코스닥'}
SECTORS = {
    '5044': '반도체', '5043': '자동차', '5048': '에너지화학', '5045': '헬스케어',
    '5064': '정보기술', '5046': '은행', '5054': '증권', '5056': '보험',
    '5049': '철강', '5052': '건설', '5057': '운송', '5051': '방송통신',
    '5063': 'K콘텐츠', '5061': '경기소비재', '5062': '필수소비재',
    '5055': '기계장비', '5065': '유틸리티',
}
STOCKS = {   # 학생은 이 부분만 자기 산업으로 교체
    '105560': 'KB금융', '055550': '신한지주', '086790': '하나금융지주',
    '316140': '우리금융지주', '024110': '기업은행', '138930': 'BNK금융지주',
    '139130': 'iM금융지주', '175330': 'JB금융지주', '323410': '카카오뱅크',
}

_now = datetime.now(KST)
_start = (_now - pd.Timedelta(days=20)).strftime('%Y%m%d')
_today = _now.strftime('%Y%m%d')
_intraday = _now.hour < CLOSE_CUTOFF_HOUR
def _trim(df):
    return df[df.index.date < _now.date()] if _intraday else df

def _index_table(codes):
    rows = []
    for code, name in codes.items():
        try:
            df = _trim(stock.get_index_ohlcv(_start, _today, code))
            rows.append({'구분': name, '종가': round(df['종가'].iloc[-1], 2),
                         '등락률(%)': round(df['종가'].pct_change().iloc[-1]*100, 2),
                         '기준일': df.index[-1].strftime('%Y-%m-%d')})
        except Exception as e:
            print(f'  WARN {name}({code}): {str(e)[:40]}')
        time.sleep(0.4)
    return pd.DataFrame(rows)

def _stock_table(codes):
    rows = []
    for code, name in codes.items():
        try:
            df = _trim(stock.get_market_ohlcv(_start, _today, code))
            last = df.iloc[-1]
            rows.append({'종목': name, '종가': f"{last['종가']:,.0f}",
                         '등락률(%)': round(last['등락률'], 2),
                         '기준일': df.index[-1].strftime('%Y-%m-%d')})
        except Exception as e:
            print(f'  WARN {name}({code}): {str(e)[:40]}')
        time.sleep(0.4)
    return pd.DataFrame(rows)

def _flow_table(date):
    want = ['개인', '외국인', '기관합계']
    rows = []
    for mkt, label in [('KOSPI', '코스피'), ('KOSDAQ', '코스닥')]:
        try:
            inv = stock.get_market_trading_value_by_investor(date, date, mkt)
            row = {'시장': label}
            for x in want:
                row[f'{x}(억)'] = f"{inv.loc[x,'순매수']/1e8:+,.0f}" if x in inv.index else '-'
            rows.append(row)
        except Exception as e:
            print(f'  WARN {label} 수급: {str(e)[:40]}')
    return pd.DataFrame(rows)

def show_report(text):
    safe = html.escape(text)
    display(HTML(f"""
<button onclick="const ta=this.nextElementSibling; ta.select(); document.execCommand('copy'); this.innerText='복사됨'; setTimeout(()=>this.innerText='전체 복사',1500);"
 style="margin:4px 0;padding:8px 16px;font-size:14px;border-radius:6px;border:1px solid #999;background:#f5f5f5;cursor:pointer;">전체 복사</button>
<textarea readonly onclick="this.select()"
 style="display:block;width:100%;height:350px;font-family:monospace;font-size:12px;white-space:pre;padding:8px;box-sizing:border-box;">{safe}</textarea>"""))

print('데이터 수집 중... (20~40초)\n')
idx_df  = _index_table(INDICES)
sec_df  = _index_table(SECTORS).sort_values('등락률(%)', ascending=False)
stk_df  = _stock_table(STOCKS)
base    = idx_df['기준일'].iloc[0] if len(idx_df) else _today
flow_df = _flow_table(base.replace('-', ''))

report  = f"# 데일리 시장 리뷰 — {base}\n\n"
report += "## 지수\n" + idx_df.to_markdown(index=False) + "\n\n"
report += "## 수급 (순매수, 억원)\n" + flow_df.to_markdown(index=False) + "\n\n"
report += "## 업종 (KRX 섹터지수, 등락률순)\n" + sec_df.to_markdown(index=False) + "\n\n"
report += "## 관심종목 (은행)\n" + stk_df.to_markdown(index=False) + "\n"

show_report(report)
```
